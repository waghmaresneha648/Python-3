```python
spam_words = ["free", "win", "winner", "cash", "urgent", "click here", "buy now", "guaranteed"]

text = input("Enter an email subject line or message: ")
text_lower = text.lower()

score = 0
reasons = []

# 1. Check for spam words
for word in spam_words:
    if word in text_lower:
        score += 2
        reasons.append(f"Contains spam phrase: '{word}'")

# 2. Check for ALL CAPS abuse
if text.isupper() and len(text) > 5:
    score += 3
    reasons.append("Entire message is in ALL CAPS")

# 3. Check for excessive exclamation marks
if text.count("!") >= 3:
    score += 2
    reasons.append("Too many exclamation marks (!)")

print("\n=== Spam Analysis ===")
if score >= 3:
    print(" Result: LIKELY SPAM / CLICKBAIT")
else:
    print(" Result: LOOKS SAFE")

print("Risk Score:", score)
if reasons:
    print("Triggers found:", ", ".join(reasons))
```
